# Spark Smoke Test
Quick check that Spark runs and can execute a simple job.

```python
from pyspark.sql import SparkSession, functions as F

spark = (
    SparkSession.builder
    .appName("spark-smoke-test")
    .master("local[*]")
    .getOrCreate()
)

df = spark.range(5).withColumn("double_id", F.col("id") * 2)
df.show()
```
